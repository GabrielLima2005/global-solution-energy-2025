```python
import pandas as pd
import matplotlib.pyplot as plt

# carregar os dados
df = pd.read_csv("consumo_simulado.csv")

print("Primeiras linhas do dataset:")
print(df.head())

# soma por hora
df["Total_kWh"] = df.iloc[:, 1:].sum(axis=1)

# gráfico simples
plt.plot(df["Hora"], df["Total_kWh"])
plt.xlabel("Hora do dia")
plt.ylabel("Consumo total (kWh)")
plt.title("Consumo de energia ao longo do dia")
plt.show()

# identificar horas de desperdício
desperdicio = df[df["Hora"] >= 20]
print("\nConsumo após 20h:")
print(desperdicio)
```
