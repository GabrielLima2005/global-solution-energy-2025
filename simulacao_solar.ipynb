```python
import matplotlib.pyplot as plt

geracao_solar = [0,0,0,1,2,3,4,5,6,5,4,3,2,1,0]  # geração solar
consumo_medio = 5  # consumo fixo

consumo_real = [consumo_medio - g for g in geracao_solar]

print("Consumo final considerando geração solar:")
print(consumo_real)

plt.plot(consumo_real)
plt.title("Consumo após energia solar")
plt.xlabel("Horas (6h às 20h)")
plt.ylabel("kWh")
plt.show()
```
